```python
import os
import shutil

def makedirs(path, remove=False):
    if os.path.isdir(path):
        if remove:
            shutil.rmtree(path)
            print("removed existing directory...")
        else:
            return
    os.makedirs(path)

def download_youtube(url, out_dir):
    try:
        import yt_dlp
    except ImportError:
        raise RuntimeError(
            "yt_dlp is not installed. Install it with 'pip install yt_dlp'."
        )

    makedirs(out_dir, remove=False)
    output_tmpl = os.path.join(out_dir, "video.%(ext)s")
    ydl_opts = {
        "format": "bestvideo[ext=mp4]+bestaudio/best[ext=mp4]/best",
        "outtmpl": output_tmpl,
        "quiet": True,
        "no_warnings": True,
    }
    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        info = ydl.extract_info(url, download=True)
        ext = info.get("ext", "mp4")
        video_path = os.path.join(out_dir, f"video.{ext}")

    if not os.path.exists(video_path):
        for fn in os.listdir(out_dir):
            if fn.startswith("video."):
                video_path = os.path.join(out_dir, fn)
                break
    return video_path


# --- EXECUTE ---
# Replace with your video URL
VIDEO_URL = "https://www.youtube.com/watch?v=tP2itcRoR5k"
CKPT_PATH = "scripts//ckpt"

# process_video_pipeline(VIDEO_URL, CKPT_PATH)
download_youtube(VIDEO_URL, "data/")
```
